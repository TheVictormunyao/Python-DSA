```python
#number of days 
days = int(input("Enter number of days: "))

seconds = days * 24 * 60 * 60

#display seconds in the days
print("The number of seconds in",days, "days is",seconds)
```

```python
#enter sphere radius
radius = float(input("Enter the radius of the sphere: "))

#cube of the radius 
radiusCubed = radius ** 3

#volume using the formula V = (4/3) * pi * r^3
volume = (4 / 3) * 3.14159 * radiusCubed

# Display the volume 
print("The volume of the sphere is:", volume)
```

```python
#Function to calculate the area of a square
def calculate_area(side):
    return side ** 2


#Function to calculate the perimeter of a square
def calculate_perimeter(side):
    return 4 * side


#enter the side length
length = float(input("Enter the side length of the square: "))

# Call the functions and store their results
area = calculate_area(length)
perimeter = calculate_perimeter(length)

# Display the results
print("The area of the square is: ", area)
print("The perimeter of the square is: ", perimeter)
```

```python
# Function to check whether the character is uppercase or lowercase
def checkCase(character):
    if character.isupper():
        return "The character is uppercase."
    elif character.islower():
        return "The character is lowercase."
    else:
        return "The input is not a letter."


#user enters a character
character = input("Enter a character: ")

# Checking the case of the character 
result = checkCase(character)

# Display the result
print(result)
```
