```python
#number of days 
days = int(input("Enter number of days: "))

seconds = days * 24 * 60 * 60

#display seconds in the days
print("The number of seconds in",days, "days is",seconds)
```

```python
#enter sphere radius
radius = float(input("Enter the radius of the sphere: "))

#cube of the radius 
radiusCubed = radius ** 3

#volume using the formula V = (4/3) * pi * r^3
volume = (4 / 3) * 3.14159 * radiusCubed

#resulting volume 
print("The volume of the sphere is:", volume)
```

```python
#Function to calculate the area of a square
def calculate_area(side):
    return side ** 2


#Function to calculate the perimeter of a square
def calculate_perimeter(side):
    return 4 * side


#enter the side length
length = float(input("Enter the side length of the square: "))

#calling functions and storing their results
area = calculate_area(length)
perimeter = calculate_perimeter(length)

#Display the results
print("The area of the square is: ", area)
print("The perimeter of the square is: ", perimeter)
```

```python
# Function to check whether the character is uppercase or lowercase
def checkCase(character):
    if character.isupper():
        return "The character is uppercase."
    elif character.islower():
        return "The character is lowercase."
    else:
        return "The input is not a letter."


#user enters a character
character = input("Enter a character: ")

# Checking the case of the character 
result = checkCase(character)

# Display the result
print(result)
```

```python
#values of x and y
x = 0
y = 20

#repeat until y becomes less than 6
while True:
    #subtract 4 from y
    y = y - 4

    #add 2 divided by y to x
    x = x + (2 / y)

    #stop loop when y is less than 6
    if y < 6:
        break

#final value of x
print("The value of x is:", x)
```

```python
#empty list to store the values
values = []

#loop to get 5 values from the user
for i in range(5):
    number = float(input("Enter a value: "))
    values.append(number)

#total of all the values
total = sum(values)

#mean average
average = total / len(values)

# Display the values and their average
print("The values entered are:", values)
print("The average is:", average)
```

```python
#Codility Challenge 7

def solution(A):
    #set to store absolute values
    distinctValues = set()
    
    #Go through every value/element in array
    for number in A:
        #Add the absolute value to set
        distinctValues.add(abs(number))
    
    #length of the number of distinct values
    return len(distinctValues)

#Example Array from challenge
A = [-5, -3, -1, 0, 3, 6]

result = solution(A)

print("Number of distinct absolute values is:", result)
```
